```python
import pandas as pd
import joblib
```

```python
model = joblib.load("../models/random_forest_model.pkl")
scaler = joblib.load("../models/scaler.pkl")
```

```python
new_house = pd.DataFrame({
    "MedInc": [5.0],
    "HouseAge": [20.0],
    "AveRooms": [6.0],
    "AveBedrms": [1.0],
    "Population": [1000.0],
    "AveOccup": [3.0],
    "Latitude": [34.0],
    "Longitude": [-118.0]
})
```

```python
new_house_scaled = scaler.transform(new_house)
```

```python
prediction = model.predict(new_house_scaled)
print("Predicted House value: ",prediction[0])
```

```python
predicted_price = prediction[0] * 100000

print(f"Predicted House Price: ${predicted_price:,.2f}")
```

```python
print("House Details:")
print(new_house)

print("\nPredicted House Price:")
print(f"${predicted_price:,.2f}")
```

```python
def predict_house_price(house_data):
    house_scaled = scaler.transform(house_data)
    prediction = model.predict(house_scaled)
    price = prediction[0] * 100000
    
    return price
```

```python
predicted_price = predict_house_price(new_house)

print(f"Predicted House Price: ${predicted_price:,.2f}")
```

```python
new_house_2 = pd.DataFrame({
    "MedInc": [3.5],
    "HouseAge": [35.0],
    "AveRooms": [5.0],
    "AveBedrms": [1.1],
    "Population": [1500.0],
    "AveOccup": [2.8],
    "Latitude": [37.5],
    "Longitude": [-122.0]
})
```
