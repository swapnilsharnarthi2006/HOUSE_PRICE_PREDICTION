```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/housing_clean.csv")
```

```python
plt.figure(figsize=(10,6))
sns.histplot(
    df["MedHouseVal"],
    bins=30,
    kde=True
)
plt.title("Distrubution of House Prices ")
plt.xlabel("Medaian House Value")
plt.ylabel("Frequency")
plt.show()
```

```python
df.hist(figsize=(15,12))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,8))
sns.boxplot(data=df)
plt.xticks(rotation=45)
plt.title("Box Plot of all Features")
plt.show()
```

```python
correlations = df.corr()
```

```python
correlations
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(
    correlations,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlations Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x = "MedInc",
    y = "MedHouseVal",
    data = df
)
plt.title("Medain Income vs Meadian House Value")
plt.xlabel("Medain Income")
plt.ylabel("Median House Value")
plt.show()
```
